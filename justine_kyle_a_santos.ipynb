```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Generating mock data (50 rows)
np.random.seed(42)
sunlight_hours = np.random.uniform(3, 10, 50)  # 50 rows of hours
# Energy = 1.5 * sunlight + noise
energy_kwh = (1.5 * sunlight_hours) + np.random.normal(0, 0.8, 50)

df = pd.DataFrame({
    'Sunlight_Hours': sunlight_hours,
    'Energy_Produced_kWh': energy_kwh
})

print(df.head())
```

```python
# Reshaping for Scikit-Learn
X = df[['Sunlight_Hours']] # Independent
y = df['Energy_Produced_kWh'] # Dependent

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
model = LinearRegression()
model.fit(X, y)
print("Model training complete.")
```

```python
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (Coefficient): {slope:.4f}")
print(f"Intercept: {intercept:.4f}")
```

```python
y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Mean Squared Error: {mse:.4f}")
print(f"R-squared Score: {r2:.4f}")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Actual Data (kWh)')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line (Predicted)')
plt.xlabel('Hours of Direct Sunlight')
plt.ylabel('Energy Produced (kWh)')
plt.title('Sunlight Hours vs. Solar Energy Production')
plt.legend()
plt.grid(True)
plt.show()
```
